```python
from modules.graph_functions import *
import matplotlib.pyplot as plt
import itertools as it
```

```python
N = 14 # number of nodes
r = 4 # number of non-starred nodes
iter_outer = 100 # number of random graph instances
```

```python
ans = [] # will store all graphs that satisfy our condition
```

```python
for i in range(iter_outer):
    adj_mat = adjacency_matrix_regular(N,r) # make a random graph 
    found = True  # var to indicate if we found a coloring that cannot be simplified then we move on to the next graph
    for state in it.combinations(np.arange(N),r): # iterate over all possible coloring
        G = generate_graph(N, adj_mat)
        G.set_rules(spread,disconnect)
        genereate_given_state(G,state)
        if not is_maximally_mixed(adj_mat, state): # exact rank failure certifies a non-maximally-mixed marginal
            found = False
            break
    if found:
        ans.append(adj_mat)
print(len(ans))
```

```python
# stores all found solutions to the file 
answers = np.zeros((len(ans),N,N))
for i in range(len(ans)):
    answers[i] = ans[i]
np.savetxt('answers_{}_{}.txt'.format(N,r),answers.reshape((len(ans),N**2)),fmt='%d')
```

Now we can access one of the solutions by index (e.g. ans[2]) and make an instance for the graph 

```python
G = generate_graph(N, ans[2]) # makes the graph from the ans list, that should always simplify 
G.set_rules(spread,disconnect) # set rules to be applied
G.plot_graph() 
```

```python
genereate_given_state(G,[0,1,2,3]) # generate some state for any qubits, here 0,1,2,3 are chosen arbitrary
```

```python
G.simplify() # simplifies the graph
G.plot_graph()
```
